```python
import pandas as pd

# --- STEP 1: Create Emission Factors ---
factors = [
    {"travel_mode": "flight", "unit": "passenger-km", "emission_factor": 0.15},
    {"travel_mode": "bus", "unit": "passenger-km", "emission_factor": 0.09},
    {"travel_mode": "car", "unit": "passenger-km", "emission_factor": 0.225},
    {"travel_mode": "motorbike", "unit": "passenger-km", "emission_factor": 0.112},
]

df_factors = pd.DataFrame(factors)
df_factors.to_csv("travel_emission_factors.csv", index=False)

print(" travel_emission_factors.csv created")
print(df_factors)

# --- STEP 2: Create Sample Travel Data ---
travel = pd.DataFrame([
    {"business_id": "B0001", "business_type": "Bakery", "date": "2024-01", "travel_mode": "flight", "distance_km": 750, "passengers": 1},
    {"business_id": "B0002", "business_type": "Restaurant", "date": "2024-01", "travel_mode": "bus", "distance_km": 500, "passengers": 2},
    {"business_id": "B0003", "business_type": "Supermarket", "date": "2024-01", "travel_mode": "car", "distance_km": 200, "passengers": 1},
    {"business_id": "B0001", "business_type": "Bakery", "date": "2024-02", "travel_mode": "motorbike", "distance_km": 50, "passengers": 1},
    {"business_id": "B0002", "business_type": "Restaurant", "date": "2024-02", "travel_mode": "flight", "distance_km": 600, "passengers": 1},
])

travel.to_csv("travel.csv", index=False)

print("\n travel.csv created")
print(travel)

# --- STEP 3: Merge and Calculate Emissions ---
df = travel.merge(df_factors, how="left", on="travel_mode")

# total passenger-km
df["passenger_km"] = df["distance_km"] * df["passengers"]

# emissions
df["emissions_kgCO2e"] = df["passenger_km"] * df["emission_factor"]

# --- STEP 4: Save Output ---
df.to_csv("travel_emissions_summary.csv", index=False)

print("\n travel_emissions_summary.csv created")
print(df.head())
```
